```python
import pandas as pd
```

```python
df = pd.read_csv("../ety_data.csv")
```

```python
# if df's subpart is 2_x, then create a new column called split with the value of x
df["split"] = df["subpart"].str.extract(r"[23]_(\d+)")

# change NaN to -1
df["split"] = df["split"].fillna(-1)

# then change subpart of any column that matches (2|3)_\d+ to 2 or 3
df["subpart"] = df["subpart"].str.extract(r"(\d+)_?\d*") # .fillna(df["subpart"])
```
