```python
import pandas as pd
import tiktoken
```

```python
encoding = tiktoken.encoding_for_model('gpt-4')
```

```python
df = pd.read_csv('etytree_large_joined_clean_v2.csv')
df
```

```python
total_tokens = 0
for i, row in df.iterrows():
    tokens = encoding.encode(row['ety_text'])
    total_tokens += len(tokens)
    # print(row['word'], len(tokens))
```

```python
total_raw_tokens = 0
for i, row in df.iterrows():
    try:
        tokens = encoding.encode(row['ety'])
        total_raw_tokens += len(tokens)
    except Exception as e:
        print(row['word'], e)
    # print(row['word'], len(tokens))
```

```python
print(f"Input cost: (${total_tokens * 5/1E6:.2f}) (batch rate)")
print(f"Input cost: (${total_raw_tokens * 5/1E6:.2f}) (batch rate)")
print(f"Prompt cost: ${len(df) * 315 * 5/1E6:.2f} (cost per prompt)")
```
