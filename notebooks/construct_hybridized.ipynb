```python
import re
# get all nested examples, which we can find by the regex {{[^}]*{{
_is_nested_re = re.compile(r"{{[^}]*{{")
def _is_nested(wikitext):
    return _is_nested_re.search(wikitext) is not None
nested = fine[fine['ety_text'].apply(_is_nested)]
```

```python
unnested = fine[fine['ety_text'].apply(lambda x: not _is_nested(x))]
```
