```python
import pandas as pd
```

```python
df = pd.read_csv('../wikitexts.csv')
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
templs = pd.read_csv('../ety_data.csv')
```

```python
print("Before: ", len(templs))
# drop Unnamed: 0
templs = templs.drop(columns=['Unnamed: 0'])
templs = templs.drop_duplicates()
print("After: ", len(templs))
```

```python
large = templs[10 <= templs['num_templates']]
large = large[large['num_templates'] <= 15]
```

```python
large
```

```python
large.rename(columns={'lang':'language', 'etymology_number': 'ety_number'}, inplace=True)
# drop split and subpart
large = large.drop(columns=['split', 'subpart'])
```

```python
# filter all rows of df that are stringy in the ety_number column
df_weird = df[~df['ety_number'].apply(lambda x: isinstance(x, int) or isinstance(x, float) or x.isnumeric())]
```

```python
df_weird
```

```python
# set type of etymology number to 0 if not present for df
# if the row cannot be coerced, drop the row
df['ety_number'] = df['ety_number'].fillna(0)
df['ety_number'] = pd.to_numeric(df['ety_number'], errors='coerce')
df = df.dropna(subset=['ety_number'])
df['ety_number'] = df['ety_number'].astype(int)
```

```python
# add "ety" and "templates" fields to df using a merge
merged_df = df.merge(large, on=['word', 'language', 'ety_number'], how='inner')
```

```python
merged_df
```

```python
# only keep word	language	ety_number	ety_text    templates
df_export = merged_df[['word', 'language', 'ety_number', 'ety', 'ety_text', 'templates', 'num_templates']]

# drop duplicates
df_export = df_export.drop_duplicates()
```

```python
df_export
```

```python
# write df_export
df_export.to_csv('etytree_large_joined.csv', index=False)
```
